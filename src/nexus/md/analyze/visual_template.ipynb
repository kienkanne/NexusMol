```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
name = "placeholder"  # Replace with your desired name

rmsd_file = f"rmsd_{name}.dat"
rmsf_file = f"rmsf_bb_{name}.dat"
ss_file = f"ss_summary{name}.dat"
PP_file = f"PP_hbvtime_{name}.dat"
BB_file = f"BB_hbvtime_{name}.dat"
PW_file = f"PW_hbvtime_{name}.dat"
PL_file = f"PL_hbvtime_{name}.dat"
pca_file = f"pca_{name}.dat"
cluster_file = f"cnumvtime_{name}.dat"
cpopvtime_file = f"cpopvtime_{name}.dat"
```

RMSD Backbone

```python
t, rmsd = np.loadtxt(rmsd_file, unpack=True)

plt.figure(figsize=(20,8))
plt.plot(t*0.1, rmsd)
plt.xlabel("Time (ns)")
plt.ylabel("RMSD (Å)")
plt.title("RMSD Protein Backbone")
plt.grid()
plt.show()
```

Secondary structure

```python
data =np.loadtxt(ss_file)
resid = data[:,0]
extended=data[:,1]
bridge=data[:,2]
three=data[:,3]
alpha=data[:,4]
pi=data[:,5]
turn=data[:,6]
bend=data[:,7]
helix = [0]*len(pi)
for i in range(len(pi)):
    b=three[i]
    c=alpha[i]
    d=pi[i]
    helix[i]=b+c+d
other=[0]*len(pi)
for i in range(len(pi)):
    e=bridge[i]
    f=turn[i]
    g=bend[i]
    other[i]=e+f+g
fig,ax=plt.subplots(3,1,sharex=True, sharey=True)
ax[0].plot(resid,extended,label='Extended')
ax[0].set_ylabel('Extended')
ax[1].plot(resid,helix,label='Helix')
ax[1].set_ylabel('Helix')
ax[2].plot(resid,other,label='Other')
ax[2].set_ylabel('Other')
fig.text(0.52,0.01,'Residues',ha='center')
fig.text(0.001,0.5,'Percentage',va='center',rotation='vertical')
#fig.ylabel('Percentage of Time')
plt.tight_layout()
```

PCA

```python
data = np.loadtxt(pca_file)
frame = data[:, 0]
pc1 = data[:, 1]
pc2 = data[:, 2]


# Normalize time for coloring
time = frame*0.002

plt.figure(figsize=(6,5))

sc = plt.scatter(pc1, pc2, c=time, cmap='viridis',
                 s=20, edgecolor='none')

cbar = plt.colorbar(sc)
cbar.set_label("Simulation Time (ns)", fontsize=12)

plt.xlabel("PC1 (Å)", fontsize=14)
plt.ylabel("PC2 (Å)", fontsize=14)
plt.title("Trajectory time evolution PCA plot", fontsize=14)

plt.xticks(fontsize=12)
plt.yticks(fontsize=12)

plt.tight_layout()
```

```python
from scipy.stats import gaussian_kde

xy = np.vstack([pc1,pc2])
z = gaussian_kde(xy)(xy)

plt.figure(figsize=(6,5))
scc = plt.scatter(pc1,pc2, c=z, s=5)

cbar = plt.colorbar(scc)
cbar.set_label("Gaussian kde", fontsize=12)

plt.xlabel('pc1',fontsize=14)
plt.ylabel('pc2', fontsize=14)
plt.title("PCA density plot", fontsize=20)

plt.show()
```

```python
cluster_data = np.loadtxt(cluster_file, skiprows=1)
clusters = cluster_data[:, 1]  # Extract just the cluster IDs
import matplotlib.colors as mcolors

# --- 2. Create a Discrete Colormap for Exactly 5 Clusters ---
# 'Set1', 'Dark2', or 'tab10' work great for distinct categorical colors.
# We slice it using .colors[:5] so it only holds exactly 5 highly distinct colors.
base_cmap = plt.get_cmap('Set1') 
custom_cmap = mcolors.ListedColormap(base_cmap.colors[:5])

# Define boundaries for the colorbar so each integer gets its own block
boundaries = np.arange(-0.5, 5.5, 1) # [-0.5, 0.5, 1.5, 2.5, 3.5, 4.5]
norm = mcolors.BoundaryNorm(boundaries, custom_cmap.N)

# --- 3. Plotting ---
plt.figure(figsize=(7, 6))

# Pass our custom cmap and norm to the scatter plot
scatter = plt.scatter(pc1, pc2, c=clusters, cmap=custom_cmap, norm=norm, s=6, alpha=0.7)

# Create the colorbar using the specific cluster integer values for ticks
cbar = plt.colorbar(scatter, ticks=np.arange(5))
cbar.set_label("Cluster ID", fontsize=12)
# Center the tick labels nicely in the middle of each color block
cbar.ax.set_yticklabels([f'Cluster {i}' for i in range(5)]) 

plt.xlabel('PC1', fontsize=14)
plt.ylabel('PC2', fontsize=14)
plt.title("Clusters Projected onto PCA Space", fontsize=16)
plt.grid(True, linestyle='--', alpha=0.3)

# Overlay cluster text centers for ultimate clarity
for cluster_id in np.unique(clusters):
    mask = (clusters == cluster_id)
    cluster_size = np.sum(mask) # Count how many frames are in this cluster
    
    center_x = np.median(pc1[mask])
    center_y = np.median(pc2[mask])
    
    # Label format: C0 (n=1500)
    label_text = f"C{int(cluster_id)}\n(n={cluster_size})"
    
    plt.text(center_x, center_y, label_text, 
             fontsize=10, fontweight='bold', color='white', ha='center',
             bbox=dict(boxstyle='round,pad=0.3', fc='black', alpha=0.5, ec='none'))

plt.show()
```

```python
import pandas as pd

# 1. Load the data 
# pandas automatically handles the whitespace separator and ignores the '#' comment character
df = pd.read_csv(cpopvtime_file, sep=r'\s+', comment='#')

# 2. Match your colors (excluding the first column, which is the X-axis)
num_clusters = df.shape[1] - 1
colors = plt.get_cmap('Set1').colors[:num_clusters]

# 3. Create the Line Plot
plt.figure(figsize=(9, 5.5))

# Extract the X-axis (Frame) by its index position (0)
x_data = df.iloc[:, 0]

# Loop through all remaining population columns by index position
for i in range(num_clusters):
    y_data = df.iloc[:, i + 1] # Columns 1, 2, 3...
    plt.plot(x_data, y_data, label=f"Cluster {i}", color=colors[i], linewidth=2)

plt.xlabel('Frame', fontsize=12)
plt.ylabel('Fractional Population', fontsize=12)
plt.title('Cluster Population Over Time', fontsize=14, fontweight='bold')

plt.legend(loc='upper left', bbox_to_anchor=(1.02, 1), title="Clusters")
plt.xlim(x_data.min(), x_data.max())
plt.ylim(-0.02, 1.02) # Extra breathing room so 0.0 and 1.0 lines aren't cut off
plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
